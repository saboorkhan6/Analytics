```python
import numpy as np
```

```python
arr=np.array([1,2,3])
print(arr.ndim)
print(arr)
```

```python
arr1=np.array([[1,2,3],[4,5,6]])
print(arr1)
print(arr1.ndim)
```

**Slicing**

```python
arr3=np.array([2,5,1,7,11,44,56,88,35,78])
print(arr3[2:8])
print(arr3[1:])
print(arr3[3:6:2])  #start,end,two steps
print(arr3[::2])  #start,end and two steps 
print(arr3[::-1])  #reverse
print(arr3[::-2]) #reverse and two steps
```

**Statistical operations**

```python
arr4=np.array([2,5,1,7,11,44,56,88,35,78])
np.max(arr4)  #maximum defualt in int value
np.min(arr4)  #minimum in int value
np.sum(arr4)  #sum in int value
np.mean(arr4)  #average in float value
np.median(arr4)  #middle value in float value
```
