```python
import seaborn as sns
import pandas as pd
```

```python
Grades={
    'Subjects':['Maths','Maths','English','English','Science','Science','Computer','Computer'],
    'Marks':[90,99,80,89,70,66,50,41],
    'Gender':["Male","Male","Male","Male","Female","Female","Female","Female"]
}
```

```python
data=pd.DataFrame(Grades)
```

```python
sns.barplot(x="Subjects",y="Marks",hue="Gender",data=data)
```
