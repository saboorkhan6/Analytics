```python
import seaborn as sns
import pandas as pd
```

```python
Grades={
    'Subjects':['Maths','Maths','English','English','Science','Science','Computer','Computer'],
    'Marks':[90,99,80,89,70,66,50,41],
    'Gender':["Male","Female","Male","Female","Male","Female","Male","Female"]
}
```

```python
data=pd.DataFrame(Grades)
```

```python
my_colors = ['#0077B6','#E65F2B']
sns.barplot(x="Subjects",y="Marks",hue="Gender",data=data,palette=my_colors)
```

**Countplot** : works in seaborn exatly like histogram in matplotlib

```python
s1=sns.load_dataset('titanic')
```

```python
s1.head(10)
```
