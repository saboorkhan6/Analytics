```python
import seaborn as sns
import pandas as pd
```

```python
Grades={
    'Subjects':['Maths','Maths','English','English','Science','Science','Computer','Computer'],
    'Marks':[90,99,80,89,70,66,50,41],
    'Gender':["Male","Female","Male","Female","Male","Female","Male","Female"]
}
```

```python
data=pd.DataFrame(Grades)
```

```python
my_colors = ['#0077B6','#E65F2B']
sns.barplot(x="Subjects",y="Marks",hue="Gender",data=data,palette=my_colors)
```

**Countplot** : works in seaborn exatly like histogram in matplotlib

```python
s1=sns.load_dataset('titanic')
```

```python
s1.head(10)
```

```python
sns.countplot(x='sex',hue='sex',data=s1,color='olive',palette='Set2',legend=False) 
# no need fort hue as it works on one part that includes two categories automatically...here i did this for adding diff colors(pallete)
#else it would have been simply :
# ns.countplot(x='sex',data=s1,color='olive')
```

**stripplot** : for individual identity,no continous data, it is not dependent on other, categorial value

```python
t=sns.load_dataset('titanic')
t
```

```python
sns.stripplot(x='survived',y='age',hue='pclass',data=t,palette='Set2')
#to show graph on more parameters..e.g here it is on the basis of alive,age,class
```

**Heatmap**: the brighter the colour..the higher the intensity

```python
b=sns.load_dataset('flights')
b
```

```python
new_data=b.pivot(index='year',columns='month',values='passengers')
sns.heatmap(new_data,annot=True,fmt='d')  #annot: annotation
```

```python
sns.pairplot(data=t,hue='survived')  #will not work in every situation
```
